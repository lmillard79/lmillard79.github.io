# Better Line Graphs for Hydrologic Data — a Python Port of Tony Ladson's R Method

**Companion notebook to:** _Better Line Graphs for Hydrologic Data — a Python Port of Tony Ladson's R Method_

**Source:** [Visualising Hydrologic Data](https://tonyladson.wordpress.com/2018/12/02/visualising-hydrologic-data/) (companion page, Hydrology and Water Resources Symposium 2018) and the concrete worked example, [Better line graphs for hydrologic data](https://tonyladson.wordpress.com/2018/08/06/better-line-graphs-for-hydrologic-data/) — Tony Ladson. R source: [gist.github.com/TonyLadson/732e3dbcb8aeaf76fd25c04f6ff246b7](https://gist.github.com/TonyLadson/732e3dbcb8aeaf76fd25c04f6ff246b7) (`Line_graph.R`).

Two lessons in one worked example: (1) a sequential palette (like ColorBrewer's "Blues") is the wrong choice for distinguishing *unordered categories* — it implies a false ranking and has poor contrast between adjacent lines; a qualitative palette designed for categorical data fixes both. (2) A legend forces the reader to look back and forth between line and key; labelling each line directly removes that step entirely. The example data (six creeks, invented names, made-up peak flows at six durations) is Ladson's own illustrative teaching data, not a real gauge record.

```python
import numpy as np
import matplotlib.pyplot as plt
print('matplotlib Dark2 colormap has', plt.cm.Dark2.N, 'colors -- same ColorBrewer palette Ladson used in R')
```

## 1. The (illustrative) data

```python
durations = [9, 12, 24, 36, 48, 72]
creeks = {
    'Brown Dog Ck': [1200, 1300, 1550, 1420, 1400, 1500],
    'Long Ck':      [100, 100, 100, 70, 60, 60],
    'Short Ck':     [700, 620, 800, 800, 820, 900],
    'Rocky Ck':     [400, 520, 500, 450, 420, 400],
    'Reedy Ck':     [490, 550, 500, 540, 510, 600],
    'Waterhole Ck': [320, 320, 490, 450, 420, 300],
}
print(f'{len(creeks)} creeks, {len(durations)} durations each')
```

## 2. Version 1 — the inappropriate choice

Ladson's R picks a sequential "Blues" palette (via `RColorBrewer::brewer.pal(8, 'Blues')`) to distinguish six unrelated creeks, plus a standard legend. This is the version to avoid, reproduced deliberately so the improved version below has something to improve on.

```python
blues = plt.cm.Blues(np.linspace(0.35, 0.95, len(creeks)))  # mimics brewer.pal(8,'Blues')[3:8]

fig, ax = plt.subplots(figsize=(7, 5))
for (name, flows), color in zip(creeks.items(), blues):
    ax.plot(durations, flows, marker='o', color=color, label=name)
ax.set_xlabel('Duration (h)')
ax.set_ylabel('Peak flow')
ax.legend(fontsize=8, title='')
ax.set_title('Version 1: sequential palette + legend (avoid this)', fontsize=11)
plt.tight_layout()
plt.savefig('../../images/2026-09_line-graphs-v1-avoid.png', dpi=150, bbox_inches='tight')
plt.show()
```

Blues that are close together in value (Rocky/Reedy/Waterhole Creeks, all in the 300-550 range) are also close together in colour — exactly where the reader most needs contrast, the palette gives the least.

## 3. Version 2 — qualitative palette, direct labelling

Same data, `Dark2` (a real qualitative/categorical ColorBrewer palette — matplotlib ships the identical one), and each line labelled directly at both ends instead of routed through a legend.

```python
dark2 = plt.cm.Dark2(np.linspace(0, 1, len(creeks)))

fig, ax = plt.subplots(figsize=(8, 5.5))
for (name, flows), color in zip(creeks.items(), dark2):
    ax.plot(durations, flows, marker='o', color=color)
    # label both ends, matching Ladson's annotate() calls at left and right
    ax.annotate(name, xy=(durations[0], flows[0]), xytext=(-8, 0),
                textcoords='offset points', ha='right', va='center', fontsize=8, color=color)
    ax.annotate(name, xy=(durations[-1], flows[-1]), xytext=(8, 0),
                textcoords='offset points', ha='left', va='center', fontsize=8, color=color)

ax.set_xlabel('Duration (h)')
ax.set_ylabel('Peak flow (cumec)')
ax.set_xlim(-15, 100)
ax.set_xticks(durations)
ax.set_title('Version 2: qualitative palette + direct labels', fontsize=11)
# minimalist finish: axis lines only, no grid, no box
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)
plt.tight_layout()
plt.savefig('../../images/2026-09_line-graphs-v2-better.png', dpi=150, bbox_inches='tight')
plt.show()
```

No legend to decode, no ambiguous adjacent blues — each creek is identifiable at a glance, coloured and named at both ends of its own line.
